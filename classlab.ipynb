```python
ages = [60, 62, 65, 70, 72, 75, 80]

demographicage = filter(lambda age: age > 65, ages)

print(list(demographicage))
```

```python
fruits = ['banana', 'apple', 'mangoes', 'watermelons']

demographic = filter(lambda x: 'a' in x, fruits)

print(list(demographic))
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance  # protected attribute

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def withdraw(self, amount):
        if amount <= self._balance:
            self._balance -= amount
        else:
            print("Insufficient funds.")

    def get_balance(self):
        return self._balance


# Usage
acc = BankAccount("Alice")
acc.deposit(1000)
acc.withdraw(300)

print(f"{acc.owner}'s Balance:", acc.get_balance())
```

```python
class animal:
    def __init__(self, name, species):
        self.name = name
        self._species = species

    def make_sound(self):
        return "Some sound"
    
class dog(animal):
    def __init__(self, name, breed):
        super().__init__(name, "Dog")
        self.breed = breed

    def make_sound(self):
        return "Woof! Woof!"   
    


dog1 = dog("Buddy", "Golden Retriever")
print(f"{dog1.name} is a {dog1.breed} and it says {dog1.make_sound()}") 
```
