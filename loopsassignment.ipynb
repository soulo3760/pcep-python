```python
-- LAMBDA FUNCTIONS is a small anonymous function that can take any number of arguments, but can only have one expression. It is a function without a name. It is also called an anonymous function.
# It can be used when we need a nameless function for a short period of time and it is defined using the keyword lambda.
# It can take any number of arguments but only one expression. The expression is executed and the result is returned.
# It is syntactically restricted to a single expression. It is a function object that is defined using the lambda keyword.
# It can be used to create small anonymous functions at runtime. It is often used as an argument to higher-order functions.
# It is used to create small, throwaway functions that are not needed elsewhere in the code.
# It is often used in conjunction with the map(), filter(), and reduce() functions.
# It is a powerful tool for functional programming in Python.
# It is a concise way to define a function in a single line of code.
# It is often used in data science and machine learning for data manipulation and transformation.
# It is a key feature of functional programming languages like Haskell and Lisp.
# It is a way to create small, anonymous functions that can be used in a variety of contexts.

# It is a powerful tool for creating small, reusable functions that can be used in a variety of contexts.
# its syntax is as follows:
# lambda arguments: expression  
# where arguments are the input parameters and expression is the output of the function.    
#example:
# def add(x, y):
#     return x + y
# add = lambda x, y: x + y
# print(add(2, 3)) # Output: 5
# show an example of a more advanced example of using lambda functions in Python in data manipulation and transformation.
# It is often used in conjunction with the map(), filter(), and reduce() functions, for instance in the following example:  
# map() applies a function to all items in an input list, filter() creates a list of elements for which a function returns true, and reduce() applies a rolling computation to sequential pairs of values in a list.
from functools import reduce
from operator import add    
from operator import mul
from operator import sub
from operator import truediv
from operator import mod
from operator import pow
from operator import itemgetter
from operator import itemgetter
from operator import attrgetter
from operator import methodcaller

#LETS NOW USE THE LAMBDA FUNCTION TO CREATE A ADVANCED FUNCTION THAT SHOWS ITS DATA MANIPULATION AND TRANSFORMATION CAPABILITIES USING REDUCE AND MAP () AS FOLLOWS:
from functools import reduce
FROM functools import map

#lets create using lambda function and map() to create a function that takes a list of numbers and returns a list of their squares.
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, numbers))
print(squares) # Output: [1, 4, 9, 16, 25]
#lets create using lambda function and filter() to create a function that takes a list of numbers and returns a list of even numbers.
numbers = [1, 2, 3, 4, 5]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens) # Output: [2, 4]
#lets create using lambda function and reduce() to create a function that takes a list of numbers and returns the sum of all numbers.
numbers = [1, 2, 3, 4, 5]
sum = reduce(lambda x, y: x + y, numbers)
print(sum) # Output: 15
#lets create using lambda function and reduce() to create a function that takes a list of numbers and returns the product of all numbers.
numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(product) # Output: 120
```
